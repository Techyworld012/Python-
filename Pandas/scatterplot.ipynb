```python
import matplotlib.pyplot as plt

# Study hours
hours = [1, 2, 3, 4, 5, 6, 7, 8]

# Exam marks
marks = [35, 40, 45, 52, 60, 68, 75, 82]

# Create scatter plot
plt.scatter(hours, marks)

# Labels and title
plt.xlabel("Study Hours")
plt.ylabel("Exam Marks")
plt.title("Study Hours vs Exam Marks")

# Display
plt.show()
```
