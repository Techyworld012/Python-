```python
import matplotlib.pyplot as plt

y = [23,34,56,78,65]
x=["Piyush","Ankit","Riddhi","Kirti","Luv"]
colors = ["red","yellow","green","blue","pink"]
plt.bar(x,y ,color = colors , edgecolor = "black" )
plt.xlabel("Names of the candidate",fontsize = 20)
plt.ylabel("pricess")
plt.title("Data 2026")
plt.show()
```
