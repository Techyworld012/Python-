```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns 

data = {"Days": [1, 2, 3, 4, 5], 
        "Sales": [100, 20, 350, 400, 500]}

data = pd.DataFrame(data)
print(data)        

sns.lineplot(data=data, x="Days", y="Sales")
plt.show()
```
